```python
import yaml
from ultralytics import YOLO

# Clean up all paths in data.yaml to prevent relative path errors
yaml_path = '/content/data.yaml'
with open(yaml_path, 'r') as f:
    data = yaml.safe_load(f)

data['train'] = '/content/train/images'
data['val'] = '/content/train/images'
data['test'] = '/content/train/images'  # Fallback to train if test folder doesn't exist

with open(yaml_path, 'w') as f:
    yaml.safe_dump(data, f, default_flow_style=False)

print("Configured data.yaml:")
print(data)

# Start training
model = YOLO("yolov8n.pt")
model.train(
    data=yaml_path,
    epochs=100,
    imgsz=640
)
```
